```python
import os
import pandas as pd
from tqdm import tqdm
```

```python
in_dir = '../../results/01_process_access/13_motif_matching'
out_dir = '../../results/01_process_access/14_get_mpbs'

os.makedirs(out_dir, exist_ok=True)
```

```python
for cell_type in ['HepG2', 'K562']:
    # Create output directory for each cell type if it doesn't exist
    os.makedirs(f"{out_dir}/{cell_type}", exist_ok=True)
    
    df = pd.read_csv(f'{in_dir}/{cell_type}_accessatac_peaks.filtered_mpbs.bed', sep='\t', header=None)
    df.columns = ['chrom', 'start', 'end', 'name', 'score', 'strand', 'signalValue']
    df = df[['chrom', 'start', 'end', 'name', 'score', 'strand']]
    
    tfs = df['name'].unique()
    
    for tf in tqdm(tfs):
        df_tf = df[df['name'] == tf]
        
        # Sanitize a string to make it safe for use as a filename.
        tf = tf.replace(" ", "_").replace("/", "_").replace("(", "").replace(")", "")
        df_tf.to_csv(f'{out_dir}/{cell_type}/{tf}.bed', sep='\t', index=False, header=False)
```
