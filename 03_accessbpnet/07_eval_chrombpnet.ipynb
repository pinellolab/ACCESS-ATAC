```python
import os
import json
import seaborn as sns
import pandas as pd
```

```python
in_dir = "../../results/03_accessbpnet/05_run_chrombpnet"
out_dir = "../../results/03_accessbpnet/07_eval_chrombpnet"
os.makedirs(out_dir, exist_ok=True)
```

```python
df_list = []
for cell_type in ['K562', 'HepG2']:
    for assay in ['ATAC', 'ACCESS']:
        print(f"Processing {cell_type} {assay}")
        # read a JSON file
        with open(f"{in_dir}/{cell_type}/{assay}/evaluation/chrombpnet_metrics.json", "r") as f:
            metrics = json.load(f)

        counts_metrics = pd.DataFrame(metrics["counts_metrics"]).transpose()
        profile_metrics = pd.DataFrame(metrics["profile_metrics"]).transpose()
        metrics = pd.concat([counts_metrics, profile_metrics], axis=1)
        metrics['cell_type'] = cell_type
        metrics['assay'] = assay

        df_list.append(metrics)

df = pd.concat(df_list, axis=0)
```

```python
df
```

```python
ax = sns.barplot(data=df, x='cell_type', y='spearmanr', hue='assay')
ax.set_ylabel('Spearman R')
#ax.set_ylim(0.7, 0.85)
```

```python
ax = sns.barplot(data=df, x='cell_type', 
                 y='pearsonr', hue='assay')
ax.set_ylabel('Pearson R')
#ax.set_ylim(0.7, 0.85)
```

```python
ax = sns.barplot(data=df, x='cell_type', 
                 y='median_jsd', hue='assay')
ax.set_ylabel('Median JSD')
ax.set_ylim(0.3, 0.7)
```

```python
df.to_csv(f'{out_dir}/results.csv', index=False)
```
