```r
atac_obs <- DataTrack(range = "../../results/03_accessbpnet/02_bam2bw/K562_atac.bw", 
                   genome = "hg38", type = "h", 
                   name = "ATAC", 
                   col = "#238b45",
                   ylim = c(0, 5))

atac_pred <- DataTrack(range = "../../results/03_accessbpnet/09_chrombpnet_pred/K562/ATAC/K562_chrombpnet.bw", 
                   genome = "hg38", type = "h", 
                   name = "ATAC", 
                   col = "#238b45",
                   ylim = c(0, 1))

atac_contribution <- DataTrack(range = "../../results/03_accessbpnet/06_chrombpnet_contribs/K562/ATAC/K562.counts_scores.bw", 
                   genome = "hg38", type = "h", 
                   name = "ATAC", 
                   col = "#238b45",
                   ylim = c(-0.01, 0.02))

access_obs <- DataTrack(range = "../../results/03_accessbpnet/02_bam2bw/K562_ACCESS.bw", 
                 genome = "hg38", type = "h", 
                 name = "ACCESS", 
                 col = "#6a51a3",
                 ylim = c(0, 9))

access_pred <- DataTrack(range = "../../results/03_accessbpnet/09_chrombpnet_pred/K562/ACCESS/K562_chrombpnet.bw", 
                 genome = "hg38", type = "h", 
                 name = "ACCESS", 
                 col = "#6a51a3",
                 ylim = c(0, 3.5))

access_contribution <- DataTrack(range = "../../results/03_accessbpnet/06_chrombpnet_contribs/K562/ACCESS/K562.counts_scores.bw", 
                   genome = "hg38", type = "h", 
                   name = "ACCESS", 
                   col = "#6a51a3",
                   ylim = c(-0.01, 0.02))

ctcf <- DataTrack(range = "../../data/encode_chipseq/K562/Bigwig/CTCF_ENCFF682MFJ.bigWig", 
                 genome = "hg38", 
                 type = "h", 
                 name = "ChIP-seq",              
                 #lwd   = 4,
                 #fill = "#ff7f00",
                 col = "#ff7f00",
                 ylim = c(0, 70))
```

```r
options(repr.plot.height = 10, repr.plot.width = 8)

# chr1:58,841,417-58,841,922

plotTracks(list(access_obs, access_pred, access_contribution, atac_obs, atac_pred, atac_contribution, ctcf), 
           title.width = 0.5, showTitle = TRUE, 
           chromosome = "chr1",
           from = 58841300, to = 58842000,
            sizes = c(0.2, 0.3, 0.3, 0.5, 0.3, 0.3, 0.5, 0.3), 
           transcriptAnnotation = "symbol", 
           background.title = "white", 
           col.border.title = "transparent", 
           lwd.border.title = "transparent", 
           col.axis = "black", 
           fontcolor.legend = "black",
           innerMargin = 3
          )
```
