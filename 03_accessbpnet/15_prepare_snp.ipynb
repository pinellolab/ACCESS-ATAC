```python
import os
import pandas as pd
import seaborn as sns
```

```python
in_dir = "../../data/Liver_caQTL"
out_dir = "../../results/03_accessbpnet/15_prepare_snp"

# create output directory if it doesn't exist
if not os.path.exists(out_dir):
    os.makedirs(out_dir)
```

```python
df_bg = pd.read_csv(f'{in_dir}/background_variants_overlapping_non-caPeaks.bed',
                    sep="\t")
df_bg.columns = ["chr", "start", "end", "variant_id", "peak"]
df_bg["allele1"] = df_bg["variant_id"].str.split(":").str[2]
df_bg["allele2"] = df_bg["variant_id"].str.split(":").str[3]
df_bg["pos"] = df_bg["end"]
df_bg = df_bg[["chr", "pos", "allele1", "allele2", "variant_id"]]

# make sure reference and alternate allele are just one character
# remove rows where REF or ALT are not single characters
df_bg = df_bg[df_bg["allele1"].str.len() == 1]
df_bg = df_bg[df_bg["allele2"].str.len() == 1]
df_bg["true_label"] = 0
df_bg = df_bg[["chr", "pos", "allele1", "allele2", "true_label"]]
```

```python
df_bg.head()
```

```python
df_true = pd.read_csv(f'{in_dir}/caQTL_variants_overlappingPeaks_LD-r2-0.8_withLead.bed',
                      sep="\t")
```

```python
df_true.head()
```

```python
sns.kdeplot(data=df_true, x='proxy_caQTL_beta', fill=True)
```

```python
df_true = df_true[["proxy_ID"]]
df_true["chr"] = df_true["proxy_ID"].str.split(":").str[0]
df_true["pos"] = df_true["proxy_ID"].str.split(":").str[1]
df_true["allele1"] = df_true["proxy_ID"].str.split(":").str[2]
df_true["allele2"] = df_true["proxy_ID"].str.split(":").str[3]
df_true = df_true[df_true["allele1"].str.len() == 1]
df_true = df_true[df_true["allele2"].str.len() == 1]
df_true["true_label"] = 1
df_true = df_true[["chr", "pos", "allele1", "allele2", "true_label"]]
```

```python
df_true.head()
```

```python
df_var = pd.concat([df_bg, df_true], ignore_index=True)
```

```python
df_var.head()
```

```python
df_var["variant_id"] = df_var["chr"] + ":" + df_var["pos"].astype(str) + ":" + df_var["allele1"] + ":" + df_var["allele2"] + ":" + df_var["true_label"].astype(str)
```

```python
df_var = df_var[["chr", "pos", "allele1", "allele2", "variant_id"]]
```

```python
df_var.head()
```

```python
df_var.to_csv(f'{out_dir}/snp_variants.tsv', 
             index=False, sep="\t", header=False)
```
