```r
in_dir = '../../results/04_single_cell_access_atac/25_annotation_by_marker_score'
out_dir = '../../results/04_single_cell_access_atac/28_get_peaks'

# create output directory
if (!dir.exists(out_dir)) {
  dir.create(out_dir, recursive = TRUE)
}
```

```r
obj <- readRDS(glue::glue("{in_dir}/obj.rds"))
```

```r
obj
```

```r
peaks <- row.names(obj)
```

```r
df <- data.frame(peaks = peaks)
```

```r
df$chr <- stringr::str_split_fixed(df$peaks, "-", 3)[,1]
df$start <- stringr::str_split_fixed(df$peaks, "-", 3)[,2]
df$end <- stringr::str_split_fixed(df$peaks, "-", 3)[,3]
df <- df[,c("chr", "start", "end")]
```

```r
head(df)
```

```r
write.table(df, file = glue::glue("{out_dir}/peaks.bed"), sep = "\t",
            quote = FALSE, row.names = FALSE, col.names = FALSE)
```
