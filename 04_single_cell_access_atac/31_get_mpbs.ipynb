```python
import os
import pandas as pd
from tqdm import tqdm
```

```python
in_dir = '../../results/04_single_cell_access_atac/29_motif_matching'
out_dir = '../../results/04_single_cell_access_atac/31_get_mpbs'

os.makedirs(out_dir, exist_ok=True)
```

```python
df = pd.read_csv(f'{in_dir}/peaks_mpbs.bed', sep='\t', header=None)
df.columns = ['chrom', 'start', 'end', 'name', 'score', 'strand', 'signalValue']
df = df[['chrom', 'start', 'end', 'name', 'score', 'strand']]
    
tfs = df['name'].unique()
    
for tf in tqdm(tfs):
    df_tf = df[df['name'] == tf]
        
    # Sanitize a string to make it safe for use as a filename.
    tf = tf.replace(" ", "_").replace("/", "_").replace("(", "").replace(")", "")
    df_tf.to_csv(f'{out_dir}/{tf}.bed', sep='\t', index=False, header=False)
```
